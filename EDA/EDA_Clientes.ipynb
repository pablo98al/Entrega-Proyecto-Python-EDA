```python
df_clientes[['Kidhome', 'Teenhome']] = df_clientes[['Kidhome', 'Teenhome']].astype('category')
df_clientes.dtypes
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

# Fila 1 - Visualizaciones anteriores
sns.histplot(df_clientes['Income'], bins=40, ax=axes[0, 0], color='steelblue', kde=True)
axes[0, 0].set_title('Distribución de Income')

sns.histplot(df_clientes['NumWebVisitsMonth'], bins=30, ax=axes[0, 1], color='coral', kde=True)
axes[0, 1].set_title('Visitas web mensuales')

sns.countplot(x='Kidhome', data=df_clientes, ax=axes[0, 2], palette='Set2')
axes[0, 2].set_title('Número de niños en casa')

# Fila 2 - Nuevas visualizaciones
sns.countplot(x='Teenhome', data=df_clientes, ax=axes[1, 0], palette='Set2')
axes[1, 0].set_title('Número de adolescentes en casa')

kids_data = df_clientes[['Kidhome', 'Teenhome']].apply(pd.Series.value_counts).T
kids_data.plot(kind='bar', ax=axes[1, 1], color=['#4CAF50', '#FF9800', '#2196F3'])
axes[1, 1].set_title('Kidhome vs Teenhome')
axes[1, 1].set_xlabel('')
axes[1, 1].legend(['0', '1', '2'])
axes[1, 1].tick_params(axis='x', rotation=0)

axes[1, 2].set_visible(False)  # Espacio vacío

plt.tight_layout()
plt.show()
```

- Income — Distribución completamente plana (uniforme)
    Esto es inusual. Un ingreso real nunca se distribuye así — normalmente tendría forma de campana.

- NumWebVisitsMonth
    Hay dos picos muy marcados: uno en 1-2 visitas (2.600 clientes) y otro en 32 visitas, con el resto distribuido igual entre medias. Es extraño que haya tantos justo en 32 visitas.

- Kidhome y  teenhome — Distribución perfectamente uniforme
    Los tres valores (0, 1 y 2 niños) tienen exactamente el mismo número de clientes. 

```python
df_clientes['Dt_Customer'] = pd.to_datetime(df_clientes['Dt_Customer'])
altas = df_clientes.groupby(df_clientes['Dt_Customer'].dt.to_period('M')).size()
altas.index = altas.index.astype(str)

plt.figure(figsize=(14, 4))
altas.plot(color='steelblue', marker='o', markersize=3)
plt.title('Nuevas altas de clientes por mes')
plt.ylabel('Número de clientes')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.boxplot(x='Kidhome', y='Income', data=df_clientes, ax=axes[0], palette='Set2')
axes[0].set_title('Income por número de niños')

sns.boxplot(x='Teenhome', y='Income', data=df_clientes, ax=axes[1], palette='Set2')
axes[1].set_title('Income por número de adolescentes')

plt.tight_layout()
plt.show()
```

Las tres cajas en ambos casos son casi iguales. No hay ninguna diferencia significativa en el ingreso según el número de hijos o adolescentes en casa. Esto es inisual.
