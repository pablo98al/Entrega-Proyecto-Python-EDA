Definimos una variable con las columnas que queremos pasar a float

Creo un bucle for para cambiar las , por .  ya que en el resultado de la tabla me salía NA esas columnas al hacer el cambio de tipo de dato

Reemplazar los meses en español por inglés, ya que si no, no se puede hacer el cambio de dato a la columna

Distribución de cada variable numerica según target

```python
num_cols = ['age', 'duration', 'campaign', 'pdays', 'previous',
            'emp.var.rate', 'cons.price.idx', 'cons.conf.idx', 'euribor3m', 'nr.employed']

fig, axes = plt.subplots(2, 5, figsize=(22, 8))
for ax, col in zip(axes.flatten(), num_cols):
    sns.boxplot(x='y', y=col, data=df_bank, ax=ax, palette='Set2')
    ax.set_title(col)
plt.suptitle('Variables numéricas vs Target', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

```python
# lOS QUE SE SUBSCRIBIERON TUVIERON UNA DURACIÓN DE ULTIMA INTERACCION 'DURATION' MÁS ALTA.
# euribor3m: La tasa de interés de referencia a tres meses. Los que dijeron "yes" tienen valores mucho más bajos (~1.3 vs ~4.8). Tiene sentido: cuando los tipos de interés son bajos, los depósitos a plazo son más atractivos.
# emp.var.rate y nr.employed — Mismo patrón que euribor3m: valores más bajos en los "yes". Las tres variables macroeconómicas están muy correlacionadas entre sí.
# previous — Los que suscribieron habían sido contactados más veces en campañas anteriores (mediana ~1 vs 0). Clientes con historial de contacto previo son mejores candidatos.
```

# Variables menos útiles:

- age — Las distribuciones son casi idénticas. La edad por sí sola no discrimina.
- campaign — Muy similar en ambos grupos, aunque los "no" tienen más outliers extremos.
- cons.price.idx — Diferencia mínima.
- cons.conf.idx — Ligera diferencia pero con mucho solapamiento.

## Outliers a tener en cuenta:

- duration tiene valores extremos >4000s, campaign llega a 56 contactos — ambos claramente anómalos.
- pdays está casi todo en 999 (valor que indica "no contactado"), por eso el boxplot aparece prácticamente plano.

```python
cat_cols = ['job', 'marital', 'education', 'contact', 'poutcome']

fig, axes = plt.subplots(2, 3, figsize=(20, 10))
for ax, col in zip(axes.flatten(), cat_cols):
    rates = df_bank.groupby(col)['y'].apply(lambda x: (x == 'yes').mean()).sort_values()
    rates.plot(kind='barh', ax=ax, color='steelblue')
    ax.set_title(f'Tasa de conversión por {col}')
    ax.set_xlabel('% suscripción')
axes[-1, -1].set_visible(False)
plt.tight_layout()
plt.show()
```

- job — Ocupación

    Estudiantes (31%) y jubilados (25%) son los perfiles que más suscriben, con diferencia. Tiene sentido: los jubilados buscan seguridad financiera y los estudiantes quizás reciben asesoramiento familiar.
    Blue-collar (~5%) es el perfil que menos suscribe — trabajos manuales con menor capacidad de ahorro.

- education — Educación

    Analfabetos (illiterate) destacan con ~21%, lo cual es sorprendente y probablemente se debe a que es un grupo muy pequeño (outlier estadístico).
    University degree (~14%) es el grupo más relevante por volumen, con buena tasa de conversión.
    A mayor educación básica (basic.4y, 6y, 9y), menor conversión.

- contact — Método de contacto

    Móvil dobla en conversión al teléfono fijo (~15% vs ~5%). Los contactados por móvil son más accesibles y probablemente más jóvenes.

- poutcome — Resultado campaña anterior

    Es la variable más potente de todas las categóricas: los clientes con éxito previo convierten al 65%. Si ya suscribieron antes, volverán a hacerlo.

```python
plt.figure(figsize=(12, 8))
sns.heatmap(df_bank[num_cols].corr(), annot=True, fmt='.2f',
            cmap='coolwarm', center=0)
plt.title('Correlación entre variables numéricas')
plt.tight_layout()
plt.show()
```

- Correlaciones altas positivas :

    emp.var.rate ↔ euribor3m = 0.97 y ↔ nr.employed = 0.91 — estas tres variables macroeconómicas son casi redundantes. Miden prácticamente lo mismo: el estado de la economía.
    euribor3m ↔ nr.employed = 0.95 — confirma lo anterior.
    emp.var.rate ↔ cons.price.idx = 0.78 — el empleo y la inflación también van ligados.

- Correlaciones negativas destacables:

    pdays ↔ previous = -0.59 — lógico: cuantos más días desde el último contacto, menos contactos previos recientes hay registrados.
    previous ↔ nr.employed = -0.50 y ↔ euribor3m = -0.46 — los clientes con más contactos previos fueron contactados en períodos de menor actividad económica.

    Variables independientes (casi sin correlación):

- age, duration y campaign no correlacionan con casi nada — son variables limpias e independientes.

```python
df_bank['date'] = pd.to_datetime(df_bank['date'])
ts = df_bank.groupby(df_bank['date'].dt.to_period('M'))['y'].apply(lambda x: (x == 'yes').mean())
ts.index = ts.index.astype(str)

fig, ax = plt.subplots(figsize=(14, 4))
ax.plot(ts.index, ts.values, marker='o', color='steelblue', markersize=4)

# Encontrar max y min
idx_max = ts.idxmax()
idx_min = ts.idxmin()

# Anotar máximo
ax.annotate(f'Máx: {idx_max}\n{ts[idx_max]:.1%}',
            xy=(idx_max, ts[idx_max]),
            xytext=(10, 10), textcoords='offset points',
            arrowprops=dict(arrowstyle='->', color='green'),
            color='green', fontsize=9)

# Anotar mínimo
ax.annotate(f'Mín: {idx_min}\n{ts[idx_min]:.1%}',
            xy=(idx_min, ts[idx_min]),
            xytext=(10, -25), textcoords='offset points',
            arrowprops=dict(arrowstyle='->', color='red'),
            color='red', fontsize=9)

# Marcar los puntos
ax.scatter([idx_max], [ts[idx_max]], color='green', zorder=5, s=80)
ax.scatter([idx_min], [ts[idx_min]], color='red', zorder=5, s=80)

ax.set_title('Tasa de conversión mensual')
ax.set_ylabel('% suscripción')
ax.set_xlabel('date')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Tendencia general
No hay una tendencia clara al alza ni a la baja — la tasa de conversión se mantiene bastante estable en el rango 9%-14% durante todo el período 2015-2019.
Volatilidad
Hay oscilaciones mes a mes bastante pronunciadas, lo que sugiere que factores puntuales (campañas específicas, contexto económico del momento) afectan más que la tendencia temporal.
Picos destacables

Septiembre 2016 (~14.5%) — el mes con mayor conversión de toda la serie. Podría coincidir con una campaña especialmente bien ejecutada.
Mayo 2018 (~13.5%) — segundo pico más alto.


Conclusión: El tiempo por sí solo no es un buen predictor — la tasa es relativamente estable.
